## 📝Introdução

O método de **quadrados mínimos** busca os coeficientes de uma função que represente o comportamento de dados conhecidos para realizar extrapolações e previsões. Quando tais dados contém **erros associados**, a estes podem ser modelados através da **teoria de conjuntos fuzzy**, de modo que a incerteza intrínseca aos números seja incorporada ao estimar novos valores, podendo, ainda, considerar a **interatividade** entre números fuzzy [1]. 
Para além do método de quadrados mínimos fuzzy, é possível resolver seu **problema inverso** através da **inversão direta**, obtendo, a partir de uma curva ajustada do tipo $Y = Mx +N$, os valores de causa $X_0 = \frac{Y_0 - N}{M}$ conhecendo os efeitos $Y_0$ [2]. 

## 📖Fundamentação Teórica

A **interatividade** entre números fuzzy $P$ e $Q$ pode ser modelada via Distribuições de Possibilidade Conjunta (JPD). Neste trabalho utiliza-se a família parametrizada  $\mathcal{J}_{\gamma}$, $\gamma\in[0,1]$. A partir dela, pode-se construir uma **aritmética interativa** como por exemplo a $+_{\gamma}$.


```python
import numpy as np
import matplotlib.pyplot as plt

def fill_ntrifuzzy_3D(ax, p1, p2, p3):

    n = 100

    # Parâmetros
    u = np.linspace(0, 1, n)
    v = np.linspace(0, 1, n)

    U, V = np.meshgrid(u, v)

    # Coordenadas dos vértices
    x1, y1, z1 = p1
    x2, y2, z2 = p2
    x3, y3, z3 = p3

    # Coordenadas baricêntricas
    W = 1 - U - V

    # Mantém somente a região triangular
    mask = W < 0

    X = W * x1 + U * x2 + V * x3
    Y = W * y1 + U * y2 + V * y3
    Z = W * z1 + U * z2 + V * z3

    X[mask] = np.nan
    Y[mask] = np.nan
    Z[mask] = np.nan

    ax.plot_surface(
        X,
        Y,
        Z,
        color="green",
        alpha=0.2,
        linewidth=0,
        antialiased=True
    )
```

```python
A = np.array([-1, 0, 1])
muA = np.array([0, 1, 0])

fig = plt.figure(figsize=(13, 4), dpi=200)

# JPD Gamma 1
ax1 = fig.add_subplot(131, projection='3d')

ax1.plot(A, [2,2,2], muA, color="#547757", linewidth=2)
ax1.plot([-2,-2,-2], A, muA, color="#547757", linewidth=2)

fill_ntrifuzzy_3D(ax1, [-1,2,0], [0,2,1], [1,2,0])
fill_ntrifuzzy_3D(ax1, [-2,-1,0], [-2,0,1], [-2,1,0])

ax1.plot(A, A, muA, color="#547757", linewidth=2)
ax1.plot(A, -A, muA, color="#547757", linewidth=2)

pico = [0,0,1]
fill_ntrifuzzy_3D(ax1, [-1,-1,0], pico, [1,-1,0])
fill_ntrifuzzy_3D(ax1, [1,-1,0], pico, [1,1,0])
fill_ntrifuzzy_3D(ax1, [-1,1,0], pico, [1,1,0])
fill_ntrifuzzy_3D(ax1, [-1,-1,0], pico, [-1,1,0])

ax1.set_title(r"JPD $\mathcal{J}_1$", y=1, fontsize=20)
ax1.set_xlabel("P", labelpad=-4, fontsize=14)
ax1.set_ylabel("Q", labelpad=-4, fontsize=14)
ax1.set_zlabel("Pertinência", labelpad=-4, fontsize=14)

# JPD Gamma 0.5
ax2 = fig.add_subplot(132, projection='3d')

ax2.plot(A, [2,2,2], muA, color="#547757", linewidth=2)
ax2.plot([-2,-2,-2], A, muA, color="#547757", linewidth=2)

fill_ntrifuzzy_3D(ax2, [-1,2,0], [0,2,1], [1,2,0])
fill_ntrifuzzy_3D(ax2, [-2,-1,0], [-2,0,1], [-2,1,0])

ax2.plot(A, A, muA, color="#547757", linewidth=2)
ax2.plot(A, [0,0,0], muA, color="#547757", linewidth=2)
ax2.plot([0,0,0], A, muA, color="#547757", linewidth=2)

pico = [0,0,1]
fill_ntrifuzzy_3D(ax2, [-1,0,0], pico, [-1,-1,0])
fill_ntrifuzzy_3D(ax2, [-1,-1,0], pico, [0,-1,0])
fill_ntrifuzzy_3D(ax2, [0,-1,0], pico, [1,0,0])
fill_ntrifuzzy_3D(ax2, [1,0,0], pico, [1,1,0])
fill_ntrifuzzy_3D(ax2, [1,1,0], pico, [0,1,0])
fill_ntrifuzzy_3D(ax2, [0,1,0], pico, [-1,0,0])

ax2.set_title(r"JPD $\mathcal{J}_{0.5}$", y=1, fontsize=20)
ax2.set_xlabel("P", labelpad=-4, fontsize=14)
ax2.set_ylabel("Q", labelpad=-4, fontsize=14)
ax2.set_zlabel("Pertinência", labelpad=-4, fontsize=14)

# JPD Gamma 0
ax3 = fig.add_subplot(133, projection='3d')

ax3.plot(A, [2,2,2], muA, color="#547757", linewidth=2)
ax3.plot([-2,-2,-2], A, muA, color="#547757", linewidth=2)

fill_ntrifuzzy_3D(ax3, [-1,2,0], [0,2,1], [1,2,0])
fill_ntrifuzzy_3D(ax3, [-2,-1,0], [-2,0,1], [-2,1,0])

ax3.plot(A, A, muA, color="#547757", linewidth=2)

pico = [0,0,1]
fill_ntrifuzzy_3D(ax3, [-1,-1,0], pico, [1,1,0])

ax3.set_title(r"JPD $\mathcal{J}_0$", y=1, fontsize=20)
ax3.set_xlabel("P", labelpad=-4, fontsize=14)
ax3.set_ylabel("Q", labelpad=-4, fontsize=14)
ax3.set_zlabel("Pertinência", labelpad=-4, fontsize=14)

# parâmetros de plot
for ax in [ax1, ax2, ax3]:
    ax.set_xticks(A)
    ax.set_yticks(A)
    ax.set_zticks([0,0.5,1])

    ax.tick_params(axis='x', labelsize=12, pad=-4)
    ax.tick_params(axis='y', labelsize=12, pad=-4)
    ax.tick_params(axis='z', labelsize=12, pad=-2)

    ax.view_init(elev=36, azim=320)

plt.subplots_adjust(
    left=0.0,
    right=0.86,
    bottom=0.05,
    top=0.90,
    wspace=0.08
)

fig.savefig("JPD_Gammas.png", bbox_inches="tight")

plt.show()
```


Dados números triangulares fuzzy $P = (p_1;p_2;p_3)$ e  $Q = (q_1;q_2;q_3)$, cujos $\alpha$-níveis são denotados respectivamente por $[p_\alpha^-,p_\alpha^+]$ e $ [q_\alpha^-,q_\alpha^+]$, $\forall \alpha \in [0,1]$, as operações de soma usual (+) [3], soma interativa otimista ($+_0$) [3], a divisão usual fuzzy ($\div$) [3], e a g-divisão ($\div_g$) [1], são dadas por, respectivamente.

$$
        [P+Q]^\alpha = [P]^\alpha + [Q]^\alpha = [\,p_1^\alpha + q_1^\alpha,\; p_2^\alpha + q_2^\alpha\,].
$$ 

$$
P+_0Q = \begin{cases}
((p_1+q_3) \land (p_2+q_2); p_2+q_2 ; (p_2+q_2) \lor (p_3+q_1)), \text{se diam}(P) \geq \text{diam} (Q) \\
((p_3+q_1) \land (p_2+q_2); p_2+q_2 ; (p_2+q_2) \lor (p_1+q_3)), \text{se diam}(P) \leq \text{diam}(Q)
\end{cases}.
$$

$$
    \left[\frac{P}{Q}\right]_\alpha = {[P]_\alpha}\div{[Q]_\alpha} =
    [\,p_\alpha^-,\; p_\alpha^+\,]
    \;\left[\,\tfrac1{q_\alpha^+},\; \tfrac1{q_\alpha^-}\right]
    = [\,\min S,\; \max S\,],
    \;S = \Bigl\{\,\tfrac{p_\alpha^-}{q_\alpha^-},\;\tfrac{p_\alpha^-}{q_\alpha^+},\;\tfrac{p_\alpha^+}{q_\alpha^-},\;\tfrac{p_\alpha^+}{q_\alpha^+}\Bigr\},
$$

$$
    [R]_\alpha= \left[\frac{P}{Q}\right]_\alpha=[P]_\alpha\div_g[Q]_\alpha\Leftrightarrow \begin{cases}
        (i) & [P]_\alpha=[Q]_\alpha[R]_\alpha,  \text{ ou}\\
        (ii) & [Q]_\alpha=[P]_\alpha[R]_\alpha^{-1}
    \end{cases}.
$$

Desse modo, foram obtidas as previsões e os erros de cada abordagem. Note que as curvas de calibração fuzzy obtidas nos dois casos que consideram a soma interativa são identicas, visto que estas não dependem da operação de divisão alterada nos dois casos fuzzy. Em comparação com a curva de calibração que considera a divisão não interativa, vemos que a banda de incerteza apresentada ao utilizar $+_0$ é significativamente menos. Assim, como esperado, todos os picos previstos coincidem com a previsão crisp, as previsões 1 estão contidas nas previsões 2 para cada método interativo, e nota-se que as previsões obtidas utilizando a g-divisão apresentam diâmetros menores que suas correspondentes usuais, e consequentemente, menor erro, como visto principalmente nos valores mais baixos de concentração. Percebe-se que o erro em $P_{1_\text{g-fuzzy}}$ é muito próximo do erro em $P_{\text{real}}$, indicando que a inceteza associada a esse método é muito pequena. Já o erro de $P_{2_\text{g-fuzzy}}$ é próxima ao de $P_{1_\text{fuzzy}}$, ou seja, o menor diâmetro obtido pela inversão com a divisão usual é próximo ao maior diâmetro obtido pela g-divisão. Além disso, a $P_{2_\text{fuzzy}}$ apresenta um erro muito maior que as demais e próxima a variação entre os dados, sendo a menos adequada para este problema. Ao comparar com o método não interativo, vemos que a propagração de incerteza dada pelas operações de subtração e divisão interativas aumentam significativamente o erro de previsão, com o diâmetro das previsões fuzzy sendo de até 0.5 M, e erro médio quatro vezes maior que o pior caso interativo e dez vezes maior que o caso crisp.
